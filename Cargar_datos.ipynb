Importamos las librerias

```python
from pathlib import Path
import pandas as pd

# Rutas relativas a la raíz del repositorio (este notebook vive en la raíz)
DATA_DIR = Path("data")
RUTA_XLSX = DATA_DIR / "BD_creditos.xlsx"   # archivo original entregado
RUTA_CSV = DATA_DIR / "BD_creditos.csv"     # archivo que consumirán los demás notebooks

DATA_DIR.mkdir(exist_ok=True)
print("Archivo de origen existe:", RUTA_XLSX.exists())
```

Cargamos el archivo original

```python
df = pd.read_excel(RUTA_XLSX, sheet_name="Hoja1")
print(f"Filas: {df.shape[0]:,} | Columnas: {df.shape[1]}")
```

Validación de la carga

```python
df.head()
```

```python
df.info()
```

Exportar a CSV

```python
df.to_csv(RUTA_CSV, index=False, encoding="utf-8")
print(f"Archivo guardado en: {RUTA_CSV}  ({RUTA_CSV.stat().st_size / 1024:,.0f} KB)")
```

Verificar que el CSV quedo identico al original

```python
df_check = pd.read_csv(RUTA_CSV, parse_dates=["fecha_prestamo"])

print("Misma forma:", df_check.shape == df.shape)
print("Mismas columnas:", list(df_check.columns) == list(df.columns))
print("Nulos por columna iguales:", (df_check.isna().sum() == df.isna().sum()).all())
print("Suma de capital_prestado igual:", round(df_check["capital_prestado"].sum(), 2) == round(df["capital_prestado"].sum(), 2))
```
